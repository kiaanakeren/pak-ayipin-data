```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Mengatur gaya seaborn
sns.set_style('whitegrid')

# Memuat dataset
df = pd.read_csv('Student_Depression_Dataset.csv')
print(df.head())
print(df.info())
```

```python
# Menampilkan ringkasan statistik
print(df.describe())

# Rata-rata pengeluaran di luar studi (Cost_without_study) berdasarkan gaya hidup (Living_style)
rata_biaya = df.groupby('Living_style')['Cost_without_study'].mean()
print(rata_biaya.sort_values(ascending=False))
```

```python
total_biaya_hidup = df.groupby('Living_style')['Cost_without_study'].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
plt.title('Total Pengeluaran di Luar Studi Berdasarkan Gaya Hidup')
sns.barplot(x=total_biaya_hidup.values, y=total_biaya_hidup.index, palette='Blues_r')
plt.xlabel('Total Biaya (Currency)')
plt.ylabel('Gaya Hidup (Living Style)')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))
sns.countplot(data=df, x='Current_semester', palette='viridis')
plt.title('Jumlah Mahasiswa Berdasarkan Semester Saat Ini')
plt.xlabel('Semester')
plt.ylabel('Jumlah Mahasiswa')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.scatterplot(data=df, x='Age', y='CGPA', hue='Depression', alpha=0.7)
plt.title('Hubungan Usia dengan IPK Berdasarkan Status Depresi')
plt.xlabel('Usia (Age)')
plt.ylabel('IPK (CGPA)')
plt.tight_layout()
plt.show()
```

```python
# Histogram distribusi IPK (CGPA)
plt.figure(figsize=(7, 5))
sns.histplot(df['CGPA'], bins=10, color='seagreen', kde=True)
plt.title('Distribusi IPK Mahasiswa (CGPA)')
plt.xlabel('CGPA')
plt.ylabel('Frekuensi')
plt.show()

# Boxplot untuk mendeteksi outlier pada biaya hidup berdasarkan latar belakang keluarga
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Family_background', y='Cost_without_study', palette='Set2')
plt.title('Sebaran Biaya Hidup Berdasarkan Latar Belakang Keluarga')
plt.xticks(rotation=20)
plt.xlabel('Latar Belakang Keluarga')
plt.ylabel('Cost Without Study')
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Grafik 1: Barplot Total Biaya per Living Style
sns.barplot(x=total_biaya_hidup.values, y=total_biaya_hidup.index, ax=axes[0, 0], palette='Blues_r')
axes[0, 0].set_title('Total Biaya per Gaya Hidup')

# Grafik 2: Countplot Semester Mahasiswa
sns.countplot(data=df, x='Current_semester', ax=axes[0, 1], palette='viridis')
axes[0, 1].set_title('Jumlah Mahasiswa per Semester')

# Grafik 3: Histogram Distribusi CGPA
sns.histplot(df['CGPA'], bins=10, ax=axes[1, 0], color='seagreen', kde=True)
axes[1, 0].set_title('Distribusi IPK (CGPA)')

# Grafik 4: Boxplot Cost without study per Family background
sns.boxplot(data=df, x='Family_background', y='Cost_without_study', ax=axes[1, 1], palette='Set2')
axes[1, 1].set_title('Outlier Biaya Hidup per Latar Belakang Keluarga')
axes[1, 1].tick_params(axis='x', rotation=15)

plt.tight_layout()
plt.show()
```

1. Tujuan Utama Skrip
Skrip ini bertujuan untuk menganalisis faktor-faktor yang memengaruhi kehidupan mahasiswa—mencakup kondisi akademik (IPK/CGPA, semester), kondisi finansial/pengeluaran di luar biaya studi, latar belakang keluarga, gaya hidup, hingga kaitannya dengan status kesehatan mental (depresi).

2. Alur & Hasil Analisis Berdasarkan Komponen Kode
Eksplorasi Awal Data (info, head, describe):

Berfungsi untuk mengenali struktur dataset, mengecek tipe data, mendeteksi ada atau tidaknya nilai yang hilang (missing values), serta melihat ringkasan statistik dasar (seperti rata-rata usia, sebaran IPK, dan kisaran biaya).

Analisis Finansial & Gaya Hidup (Cost_without_study & Living_style):

Bar Plot Total Pengeluaran: Mengelompokkan total biaya hidup di luar studi berdasarkan gaya hidup mahasiswa. Ini menunjukkan kategori living style mana yang menyedot pengeluaran non-akademik paling besar.

Box Plot Latar Belakang Keluarga vs Biaya Hidup: Mendeteksi variasi pengeluaran dan menemukan apakah ada nilai ekstrem (outlier) pengeluaran pada kelompok latar

Analisis Akademik & Demografi Mahasiswa:

Count Plot & Pie Chart Semester (Current_semester): Memetakan sebaran jumlah mahasiswa per semester serta menyoroti proporsi persentase mahasiswa di 5 semester teratas dengan populasi terbanyak.

Histogram Distribusi IPK (CGPA): Menampilkan bentuk kurva sebaran nilai IPK mahasiswa (lengkap dengan kurva densitas/KDE berwarna hijau) untuk melihat apakah nilai mahasiswa terkonsentrasi di angka tertentu atau menyebar merata.

Scatter Plot Usia vs IPK (Age vs CGPA dengan Hue Depression): Menganalisis korelasi antara usia dan prestasi akademik (IPK), sekaligus membedakan status depresi mahasiswa menggunakan warna titik untuk melihat apakah kelompok dengan IPK tertentu atau usia tertentu lebih rentan mengalami depresi.

Dashboard Visual Terpadu (Subplots 2x2):

Di bagian akhir, 4 visualisasi utama (Bar Chart Biaya, Count Chart Semester, Histogram IPK, dan Box Plot Outlier Biaya) digabungkan ke dalam satu bingkai grid multi-panel agar laporan atau ringkasan visual dapat dibaca secara ringkas dalam satu tampilan layar.
